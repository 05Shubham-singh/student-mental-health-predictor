```python
df=df.drop_duplicates()
```

```python
 df['Physical_Activity_Hours']=df['Physical_Activity_Hours'].clip(lower=0) #to remove negative values
```

```python
num_cols=df.select_dtypes(include='number')
num_cols.skew()
# near to 0 -> Centralized (0.01, 0.002)
# negative -> Left Skewed (-1.56)
# poistive (greater than 0) -> Right Skewed (1.256)
```

```python
top_countries=df['Country'].value_counts().index[:10].tolist()
def group_countries(country):
    if country in top_countries: 
        return country
    else:
        return 'Other'
```

```python
df['Grouped_Country']=df['Country'].apply(group_countries)
```

```python
df['Grouped_Country'].value_counts()
```

```python
from sklearn.model_selection import train_test_split
sknews_col=['Study_Hours']
other_numerical_cols=['Age','Avg_Daily_Usage_Hours', 'Daily_Unlocks',
        'Physical_Activity_Hours', 'Sleep_Hours_Per_Night']
ordinal_columns=["Stress_Level"]
normal_col=['Gender', 'Academic_Level', 'Most_Used_Platform',
       'Purpose_Of_Use','Grouped_Country']
feature_col=sknews_col+other_numerical_cols+ordinal_columns+normal_col
X=df[feature_col]
y=df['Mental_Health_Score']

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.30,random_state=42)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer, OneHotEncoder, OrdinalEncoder,StandardScaler
from sklearn.compose import ColumnTransformer

skew_pipeline=Pipeline(steps=[
    ('log_transform',FunctionTransformer(np.log1p)),
    ('scale',StandardScaler())
])

plain_numeric_pipeline=Pipeline(steps=[('scale',StandardScaler())])

ordinal_pipeline=Pipeline(steps=[
    ('orden_encoder',OrdinalEncoder(categories=[['Low','Medium','High','Very High']]))
])

nominal_pipeline=Pipeline(steps=[
    ('one_enocode',OneHotEncoder(handle_unknown='ignore'))
])

preprocessor=ColumnTransformer(transformers=[
    ("Skew_Pipeline",skew_pipeline,sknews_col),
    ("Plain_Numerical",plain_numeric_pipeline,other_numerical_cols),
    ("Ordinal_Pipeline",ordinal_pipeline,ordinal_columns),
    ("Noraml",nominal_pipeline,normal_col) 
])

preprocessor
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error

lr_pipeline=Pipeline(steps=[
    ("preprocessor",preprocessor),
    ('regression',LinearRegression())
])

lr_pipeline.fit(X_train,y_train)
lr_pred_testing=lr_pipeline.predict(X_test)
lr_predict_training=lr_pipeline.predict(X_train)

lr_r2_testing=r2_score(y_test,lr_pred_testing)
lr_r2_traning=r2_score(y_train,lr_predict_training)

lr_mean_error=mean_absolute_error(y_test,lr_pred_testing)

print(f"Accuracy of Training {lr_r2_traning}")
print(f"Accuracy of Testing {lr_r2_testing}")
print(f"MAE is {lr_mean_error}")
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_pipeline=Pipeline(steps=[
    ("preprocessor",preprocessor),
    ("Random_Forest",RandomForestRegressor(random_state=42))
])

rf_pipeline.fit(X_train,y_train)
rf_pred_testing=rf_pipeline.predict(X_test)
rf_pred_traning=rf_pipeline.predict(X_train)

rf_r2_testing=r2_score(y_test,rf_pred_testing)
rf_r2_traning=r2_score(y_train,rf_pred_traning)

rf_mae=mean_absolute_error(y_test,rf_pred_testing)

print(f"Accuracy of Training {rf_r2_traning}")
print(f"Accuracy of Testing {rf_r2_testing}")
print(f"MAE is {rf_mae}")
```

```python
from  sklearn.model_selection import RandomizedSearchCV

param_grid = {
    'Random_Forest__n_estimators'       : [100,200,300],
    'Random_Forest__max_depth'         : [5,10,15],
    'Random_Forest__min_samples_split' : [2,5,10],
    'Random_Forest__min_samples_leaf'  : [1,2,4]
}


random_search=RandomizedSearchCV(
    estimator=rf_pipeline,
    param_distributions=param_grid,
    n_iter=15,cv=5,
    scoring='r2',
    random_state=42, 
    n_jobs=-1
)

random_search.fit(X_train,y_train)
```

```python
rf_best_pipeline=random_search.best_estimator_
rf_best_pred=rf_best_pipeline.predict(X_test)
print(f"R2 score for random forest after hyperparameter tuning {r2_score(y_test,rf_best_pred)}")
print(f"MAC Score for random forest after hypertuning {mean_absolute_error(y_test,rf_best_pred)}")
```

```python
# Calculate RMSE for Linear Regression
lr_rmse = np.sqrt(mean_squared_error(y_test, lr_pred_testing))

# Calculate RMSE for default Random Forest
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred_testing))

# Calculate RMSE for tuned Random Forest
rf_tuned_preds          = random_search.best_estimator_.predict(X_test)
rf_tuned_rmse           = np.sqrt(mean_squared_error(y_test, rf_tuned_preds))
rf_tuned_training_preds = random_search.best_estimator_.predict(X_train)
r2_tuned_training       = r2_score(y_train, rf_tuned_training_preds)
rf_tuned_mae            = mean_absolute_error(y_test, rf_tuned_preds)
rf_tuned_r2             = r2_score(y_test, rf_tuned_preds)


# Create a DataFrame to consolidate results
results = pd.DataFrame({
    'Model': ['Linear Regression', 'Random Forest (default)', 'Random Forest (tuned)'],
    'R2': [lr_r2_testing, rf_r2_testing, rf_tuned_r2],
    'Training R2': [lr_r2_traning, rf_r2_traning, r2_tuned_training],
    'MAE': [lr_mean_error, rf_mae, rf_tuned_mae],
    'RMSE': [lr_rmse, rf_rmse, rf_tuned_rmse]
})

print(results)
```
